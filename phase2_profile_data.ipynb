```python
import pandas as pd
import os

DATA_DIR = "../data/raw"  # adjust if your path is different

coverage = pd.read_excel(os.path.join(DATA_DIR, "coverage-data.xlsx"))
incidence = pd.read_excel(os.path.join(DATA_DIR, "incidence-rate-data.xlsx"))
cases = pd.read_excel(os.path.join(DATA_DIR, "reported-cases-data.xlsx"))
intro = pd.read_excel(os.path.join(DATA_DIR, "vaccine-introduction-data.xlsx"))
schedule = pd.read_excel(os.path.join(DATA_DIR, "vaccine-schedule-data.xlsx"))

print("Coverage:", coverage.shape)
print("Incidence:", incidence.shape)
print("Cases:", cases.shape)
print("Introduction:", intro.shape)
print("Schedule:", schedule.shape)
```

```python
tables = {
    "coverage": coverage,
    "incidence": incidence,
    "cases": cases,
    "intro": intro,
    "schedule": schedule
}

for name, df in tables.items():
    print(f"\n===== {name.upper()} =====")
    print("Shape:", df.shape)
    print("\nColumn dtypes:")
    print(df.dtypes)
    print("\nMissing values (top 5 columns):")
    print(df.isna().sum().sort_values(ascending=False).head(5))
    print("\nDuplicate rows:", df.duplicated().sum())
```
